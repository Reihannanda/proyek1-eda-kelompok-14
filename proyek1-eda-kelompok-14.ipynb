*Keputusan Akhir Kelompok Kami:*
Karena kelompok kami tidak menemukan adanya kesalahan pada data diatas, maka kami memutuskan untuk membiarkan semua datanya tetap utuh seperti aslinya. Kelompok kami tidak melakukan revisi apapun dan kelompok kami langsung lanjut ke tahap berikutnya.

```python
import pandas as pd
import numpy as np

# Mengubah format angka ilmiah e+03 menjadi desimal biasa biar gampang dibaca
pd.set_option('display.float_format', lambda x: '%.2f' % x)

kolom_pilihan = ["cnt", "temp", "hum"]
ringkasan_stat = {}

for kol in kolom_pilihan:
    mean_val = df[kol].mean()
    median_val = df[kol].median()
    mode_val = df[kol].mode()[0]  # Mengambil nilai modus pertama
    
    min_val = df[kol].min()
    max_val = df[kol].max()
    range_val = max_val - min_val
    
    q1 = df[kol].quantile(0.25)
    q3 = df[kol].quantile(0.75)
    iqr_val = q3 - q1
    
    # Hanya memasukkan komponen utama tanpa Varians dan Simpangan Baku
    ringkasan_stat[kol] = {
        "Rata-rata (Mean)": mean_val,
        "Median": median_val,
        "Modus (Mode)": mode_val,
        "Minimum": min_val,
        "Maksimum": max_val,
        "Jangkauan (Range)": range_val,
        "Kuartil 1 (Q1)": q1,
        "Kuartil 3 (Q3)": q3,
        "IQR": iqr_val
    }

df_stat = pd.DataFrame(ringkasan_stat)
print("TABEL RINGKASAN STATISTIK DESKRIPTIF")
df_stat
```

**Setelah kelompok kami menghitung menggunakan kodingan di atas, ini petabel statistik kelompok kami:**
1. Jumlah Sepeda yang Disewa Harian (cnt):
   * Rata-rata & Nilai Tengah: Biasanya, dalam sehari ada sekitar *4.508 sepeda* yang laku disewa, dengan titik tengah data berkumpul di angka *4.548 sepeda*.
   * Paling Sepi & Paling Ramai: Hari paling sepi di data kami hanya tercatat *22 sepeda* yang laku disewa, sedangkan hari paling ramai bisa melonjak hingga *8.714 sepeda* dalam sehari.
   * *Angka Kumpulannya (IQR):* Banyaknya hari biasa mencatat jumlah sewaan yang berkumpul stabil di antara *3.169 sampai 5.966 sepeda*.

2. **Suhu Udara Harian (temp):**
   * Suhu udara harian rata-rata di data ini adalah *20,32 derajat*. Hari paling dingin tercatat di angka *2,42 derajat* dan hari paling panas mentok di angka *35,33 derajat*.

3. **Kelembapan Udara Harian (hum):**
   * Kelembapan udara rata-ratanya ada di angka *62,77%* (artinya kondisi harian udara di sana tergolong normal atau sedang).

**"Bandingkan rata-rata dan median: kalau jauh berbeda, apa artinya bagi bentuk sebaran data?"**
* Hasil Perbandingan pada Data Sepeda (cnt):
  Nilai rata-rata sewaan sepeda yang kelompok kami temukan adalah *4.508* dan nilai tengahnya (median) adalah *4.548*. Kedua angka ini mirip  dan hampir sama (hanya berbeda sekitar 40 sepeda saja).
* Artinya Bagi Bentuk Data:
  Karena nilai rata-rata dan mediannya hampir sama dan selisihnya sangat kecil, ini menandakan bentuk sebaran data tersebut cenderung simetris atau mendekati dengan distribusi normal.

```python
frekuensi = df["weathersit"].value_counts()
persentase = df["weathersit"].value_counts(normalize=True) * 100
df_kategorik = pd.DataFrame({
    'Frekuensi (Hari)': frekuensi,
    'Persentase (%)': persentase
})
df_kategorik.index = df_kategorik.index.map({
    1: '1: Cerah / Sedikit Berawan',
    2: '2: Kabut / Berawan Mendung',
    3: '3: Hujan Ringan / Salju / Badai'
})

df_kategorik = df_kategorik.reset_index()
print("TABEL FREKUENSI KONDISI CUACA HARIAN")
df_kategorik
```

### Kategori yang Paling Sering Muncul (Modus)

Kategori kondisi cuaca yang bertindak sebagai **Modus** dalam dataset ini adalah **Kondisi Cuaca 1 (Cerah / Sedikit Berawan)** dengan persentase dominasi yang sangat besar, yaitu mencapai **63,42%** dari keseluruhan total hari dalam data. 

Hal ini menunjukkan bahwa lingkungan wilayah operasional *Bike Sharing* ini secara umum didominasi oleh hari-hari dengan cuaca yang cerah dan bersahabat, yang mana kondisi atmosfer seperti ini sangat mendukung masyarakat untuk menggunakan sepeda sebagai moda transportasi harian.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

plt.hist(df["cnt"], bins=20, color="skyblue", edgecolor="black")

plt.title("1. Sebaran Total Penyewaan Sepeda Harian (cnt)", fontsize=11, fontweight="bold")
plt.xlabel("Total Penyewaan (sepeda)")
plt.ylabel("Frekuensi (hari)")
plt.show()
```

**Histogram Total Penyewaan Sepeda Harian:**
Kalau dilihat, bentuk grafik sewa sepeda harian kita ini **Simetris Seimbang**. Bentuknya rapi di tengah mirip gunung, yang artinya jumlah hari yang sepi banget dan hari yang ramai banget itu seimbang, kebanyakan kumpul di angka normal harian sekitar 4.500 sepeda.

```python
plt.figure(figsize=(8, 5))

plt.hist(df["registered"], bins=20, color="orange", edgecolor="black")

plt.title("2. Sebaran Penyewa Berlangganan Harian (registered)", fontsize=11, fontweight="bold")
plt.xlabel("Jumlah Pengguna Berlangganan (orang)")
plt.ylabel("Frekuensi (hari)")
plt.show()
```

**Histogram Penyewa Langganan:**
Bentuk grafik untuk pengguna langganan harian juga cenderung **Simetris Seimbang**. Bedanya, puncaknya melebar di area tengah ke kanan, membuktikan kalau pelanggan tetap BoomBikes jumlahnya selalu ramai dan stabil menggunakan sepeda hampir setiap hari.

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["casual"], color="salmon")

plt.title("3. Boxplot Penyewa Biasa / Non-Langganan (casual)", fontsize=11, fontweight="bold")
plt.xlabel("Jumlah Pengguna Biasa (orang)")
plt.show()
```

**Boxplot Penyewa Biasa:**
Di grafik kotak ini kita bisa melihat ada **banyak titik-titik bulat yang nyasar di sebelah kanan garis pagar luar**. Titik-titik itu adalah bukti nyata adanya nilai pencilan (*outliers*). Artinya, ada hari-hari libur spesial tertentu di mana jumlah orang biasa yang mendadak sewa sepeda melonjak tinggi banget di luar batas wajar harian.

```python
plt.figure(figsize=(8, 5))

map_musim = {1: "1. Spring", 2: "2. Summer", 3: "3. Fall", 4: "4. Winter"}
df_copy = df.copy()
df_copy["nama_musim"] = df_copy["season"].map(map_musim)
rerata_musim = df_copy.groupby("nama_musim")["cnt"].mean()

plt.bar(rerata_musim.index, rerata_musim.values, color=["lightgreen", "gold", "orange", "lightblue"], edgecolor="black")

plt.title("4. Rata-rata Penyewaan Sepeda per Musim (season)", fontsize=11, fontweight="bold")
plt.xlabel("Musim")
plt.ylabel("Rata-rata Penyewaan (sepeda)")
plt.show()
```

**Grafik Batang Rata-rata per Musim:**
Dari tiang batang warna-warni ini, kelihatan jelas kalau **Musim Gugur (Fall)** adalah waktu paling juara dan favorit bagi orang-orang buat sepedaan (rata-rata 5.644 sepeda/hari). Sedangkan **Musim Dingin (Winter)** menjadi waktu yang paling sepi (rata-rata cuma 2.608 sepeda/hari) karena udaranya bikin orang malas keluar rumah.

```python
plt.figure(figsize=(7, 5))

plt.scatter(df["temp"], df["cnt"], alpha=0.5, color="purple")

plt.title("Scatter Plot (Nilai Plus): Hubungan Suhu vs Total Penyewaan", fontsize=11, fontweight="bold")
plt.xlabel("Suhu Udara Aktual (Derajat Celsius)")
plt.ylabel("Total Penyewaan (sepeda)")
plt.grid(True)
plt.show()
```

**Scatter Plot Hubungan Suhu dengan Total Penyewaan:**
Grafik titik-titik ungu ini menunjukkan arah yang kompak **Miring Naik ke Kanan**. Ceritanya gampang banget dibaca: semakin hangat dan pas suhu udaranya (bergerak ke kanan), maka jumlah total sepeda yang laku disewa juga bakal ikutan naik drastis (bergerak ke atas). Ini bukti kalau minat orang naik sepeda sangat tergantung sama kondisi kehangatan cuaca hari itu.
